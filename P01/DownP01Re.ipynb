````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
# Load the dataset
file_path = '/content/Practical 1 research_student (1).xlsx'#change this if required.
df = pd.read_excel(file_path)

# Display the first 5 rows
display(df.head())
````

### 4. Data Preprocessing

In this step, we will:
1. Check for missing values in each column.
2. Generate initial statistics for the numerical and categorical columns.
3. Identify the dimensions (rows and columns) of the dataset.
4. Provide a description of the variables.

````python
# 1. Check for missing values
print("Missing Values per Column:")
display(df.isnull().sum())

# 2. Initial statistics
print("\nInitial Statistics (Numerical):")
display(df.describe())

# 3. Check dimensions
print(f"\nDimensions of the Data Frame: {df.shape}")
````

````python
# 4. Variable Types and Descriptions
print("Variable Info:")
display(df.info())
````

### 5. Data Formatting and Data Normalization

In this step, we will:
1. Clean the dataset by removing empty rows.
2. Verify and convert data types where necessary.
3. Summarize the variables by their finalized types.

````python
# Remove rows where all or most values are missing (cleaning up the trailing empty rows)
df.dropna(how='all', inplace=True)

# Handling specific type conversions if any categorical columns are stored as numeric or vice versa
# Based on df.info(), the types look mostly correct, but let's ensure 'Ever Back' is an integer after cleaning
df['Ever Back'] = df['Ever Back'].fillna(0).astype(int)

print("Updated Data Types:")
display(df.dtypes)

print("\nSummary of variable types:")
num_vars = df.select_dtypes(include=['number']).columns.tolist()
cat_vars = df.select_dtypes(include=['object']).columns.tolist()
print(f"Numerical variables: {len(num_vars)}")
print(f"Categorical variables: {len(cat_vars)}")
````

````python
display(df.isnull().sum())
````

### Note on Type Conversion
If a column is not in the correct data type, we use `astype()` or specific pandas conversion functions.

Example:
```python
# 1. Convert to numeric, forcing errors to NaN (useful for messy data)
df['Column_Name'] = pd.to_numeric(df['Column_Name'], errors='coerce')

# 2. Convert to integer after handling NaNs
df['Column_Name'] = df['Column_Name'].fillna(0).astype(int)

# 3. Convert to categorical string
df['Column_Name'] = df['Column_Name'].astype(str)
```

### 5. (Continued) Data Normalization

Normalization rescales numerical values to a range between 0 and 1. This is useful when different columns have different scales (e.g., Marks vs GPAs).

````python
# Identify numerical columns for normalization
cols_to_normalize = ['Marks[10th]', 'Marks[12th]', 'GPA 1', 'CGPA']

# Min-Max Normalization: (x - min) / (max - min)
for col in cols_to_normalize:
    df[col + '_normalized'] = (df[col] - df[col].min()) / (df[col].max() - df[col].min())

print("First 5 rows with normalized columns:")
# Fixed the display index syntax
columns_to_show = cols_to_normalize + [col + '_normalized' for col in cols_to_normalize]
display(df[columns_to_show].head())
````

### 6. Turn Categorical Variables into Quantitative Variables

We use One-Hot Encoding to create binary columns for each category.

````python
# 1. On current dataset
# Let's encode 'Branch' and 'Gender'
df_encoded = pd.get_dummies(df, columns=['Branch', 'Gender'], prefix=['Br', 'Gen'])

print("Encoded DataFrame Shape:", df_encoded.shape)
display(df_encoded.filter(like='Br_').head())

# 2. What if it can't be done directly (e.g., if columns were already numerical codes)?
# You would first map them to strings or use LabelEncoder
# Example of manual mapping if 'Gender' was 1 and 2:
# df['Gender'] = df['Gender'].map({1: 'Male', 2: 'Female'})
# Then apply get_dummies as shown above.
````

````python
# Identify the row(s) where Rank is present but Branch is missing
discrepancy_row = df[df['Rank'].notnull() & df['Branch'].isnull()]
print("Row(s) causing the count discrepancy:")
display(discrepancy_row)
````
