### 5. Data Formatting and Normalization
We will check the data types and perform any necessary type conversions. For example, ensuring 'Embarked' is treated correctly and handling potential missing values before conversion.

```python
# Check data types
print("Current Data Types:")
print(df.dtypes)

# Convert 'Survived' and 'Pclass' to categorical if needed for analysis,
# but for the assignment we ensure numeric types are correct.
# Example: Convert Embarked to category type
df['Embarked'] = df['Embarked'].astype('category')

# Handling missing values for Age by filling with median to allow for further steps
df['Age'] = df['Age'].fillna(df['Age'].median())

# Drop Cabin due to high missing values
df.drop(columns=['Cabin'], inplace=True)

print("\nUpdated Data Types:")
print(df.dtypes)
```

### 6. Turn Categorical Variables into Quantitative Variables
We will use Label Encoding or One-Hot Encoding for 'Sex' and 'Embarked'.

```python
# Using map for binary variable 'Sex'
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# Using get_dummies for 'Embarked'
df = pd.get_dummies(df, columns=['Embarked'], drop_first=True)

print("Dataframe after converting categorical variables:")
display(df.head())
```

```python
# Data Normalization: Scaling numeric features to a range of [0, 1]
# This is useful for many machine learning algorithms.
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])

print("Data after Min-Max Normalization on Age and Fare:")
display(df[['Age', 'Fare']].head())
```

### Comparison: Normalization vs. Standardization
- **Normalization (Min-Max):** Rescales to [0, 1]. Best when features have different scales but finite bounds.
- **Standardization (StandardScaler):** Centers data around mean 0 with unit variance. Best when data follows a Gaussian distribution or has outliers.

```python
from sklearn.preprocessing import StandardScaler

# Let's create a copy to demonstrate standardization without losing our normalized data
df_standardized = df.copy()
std_scaler = StandardScaler()

# Standardizing Age and Fare
df_standardized[['Age', 'Fare']] = std_scaler.fit_transform(df_standardized[['Age', 'Fare']])

print("Data after Standardization (Mean ≈ 0, Std ≈ 1):")
display(df_standardized[['Age', 'Fare']].describe().round(4))
```

### When to use which?

| Feature | Standardization (StandardScaler) | Normalization (Min-Max) |
| :--- | :--- | :--- |
| **Best For** | Algorithms that assume Gaussian (Normal) distribution (Linear Regression, Logistic Regression, LDA). | Algorithms that don't make distribution assumptions (KNN, Neural Networks). |
| **Outliers** | **Robust.** Outliers have less impact on the final range of most data points. | **Sensitive.** A single outlier can 'squash' all other values into a tiny range. |
| **Range** | Not bounded (usually -3 to 3, but can be anything). | Strictly bounded [0, 1] or [-1, 1]. |
| **Use Case Example** | Image processing or when you have extreme 'Fare' values that shouldn't break the model. | Algorithms where 'distance' or 'gradient' size matters strictly, like K-Nearest Neighbors or K-Means clustering. |
